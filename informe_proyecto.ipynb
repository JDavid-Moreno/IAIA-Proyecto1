## 2. Construccion del Grafo Físico y Heuristica (station_graph.py)

Para guiar la busqueda de manera inteligente, necesitamos una estimacion de que tan lejos estamos del destino (heuristica). Aqui creamos un grafo que representa la red fisica de troncales de Transmilenio y sus conexiones a pie (transbordos fisicos).

Usamos la clase `Graph` para modelar las estaciones como nodos y las conexiones como aristas. Sin embargo, no todas las conexiones tienen el mismo costo real: ademas de la estructura fisica, definimos pesos especiales para ciertos tramos donde el desplazamiento es mas costoso o mas rapido en la practica. Esto se hace mediante la funcion `edge_weight()`, que reemplaza el peso base (`EDGE_WEIGHT = 3`) en aristas clave como:
- `Calle 72` - `Polo`
- `Calle 100` - `La Castellana`
- `Heroes` - `Polo`
- `NQS Calle 75 - Zona M` - `San Martin`
- `Avenida El Dorado` - `Ciudad Universitaria - Loteria de Bogota`
- `Avenida Jimenez` - `San Victorino`
- `Tygua - San Jose` - `Bicentenario`
- `Centro Memoria` - `Universidades`
- `Portal Americas` - `Biblioteca Tintal`

Estos ajustes permiten que la aproximación sea mas realista y mejorar la calidad de la heuristica utilizada por la busqueda.

Además, para conectar las distintas troncales y modelar los transbordos entre corredores, se utilizó la lista `inter_corridor_connections`. Esta estructura agrega aristas entre estaciones clave que no pertenecen al mismo corredor, como `Heroes - Polo`, `Calle 72 - Polo`, `NQS Calle 75 - Zona M - San Martin`, `Avenida El Dorado - Ciudad Universitaria - Loteria de Bogota`, `Tygua - San Jose - Bicentenario`, `Avenida Jimenez - San Victorino`, entre otras. ## 3. Algoritmo de Busqueda A* (faster_route.py)

Este es la clase principal del proyecto. ESta basada en el Laboratorio 1, de este se reutilizaron varias clases y metodos, cambiando `problem` el cual Consiste en:
1. **Estados:** Representados por una tupla `(estacion_actual, ruta_actual)`.
2. **Acciones:** Bajar de un bus, subirse a una nueva ruta, o moverse a la siguiente estacion de la ruta actual.
3. **Costo de Acciones:** Evaluamos el tiempo que toma viajar entre estaciones (`TRAVEL_TIME`), detenerse (`DWELL_TIME`) y hacer transbordos (`TRANSFER_TIME`).
4. **Heuristica:** La funcion `heuristic` utiliza las distancias precalculadas por el grafo fisico para orientar la busqueda eficientemente.

La funcion `best_first_search` utiliza una frontera (`Frontier`) para explorar primero los caminos mas prometedores. Al finalizar, imprimira las indicaciones paso a paso (como subir a un bus y donde hacer transbordo) y el tiempo total. 

```python
import heapq

class Node:
    def __init__(self, state, parent=None, action=None, cost=0):
        self.state = state
        self.parent = parent
        self.action = action
        self.cost = cost

    def __str__(self):
        return f"Node(state={self.state}, action={self.action}, cost={self.cost})"

    def representation(self):
        return self.state


class Frontier:
    def __init__(self, evaluation_function):
        self.evaluation_function = evaluation_function
        self.frontier_node = []
        self.count = 0

    def is_empty(self):
        return len(self.frontier_node) == 0

    def pop(self):
        if self.is_empty():
            raise Exception("empty frontier")
        delete = heapq.heappop(self.frontier_node)
        return delete[2]

    def top(self):
        if self.is_empty():
            raise Exception("empty frontier")
        return self.frontier_node[0][2]

    def add(self, node):
        value = self.evaluation_function(node)
        heapq.heappush(self.frontier_node, (value, self.count, node))
        self.count += 1


class Problem:
    def __init__(self, initial_station, goal_station, routes, graph_instance):
        self.initial_station = initial_station
        self.goal_station = goal_station
        self.routes = routes
        self.routes_by_station = build_routes_by_station(routes)
        self.initial_state = (initial_station, None)
        self.distances_to_goal = graph_instance.shortest_distances_from(goal_station)

    def is_goal(self, state):
        return state[0] == self.goal_station

    def result_actions(self, state):
        station, current_route = state
        new_states = []

        if current_route is None:
            for route_id in self.routes_by_station.get(station, []):
                new_state = (station, route_id)
                new_states.append(new_state)
        else:
            route_stations = self.routes[current_route]["stations"]
            index = route_stations.index(station)

            if index + 1 < len(route_stations):
                next_station = route_stations[index + 1]
                new_state = (next_station, current_route)
                new_states.append(new_state)

            new_state = (station, None)
            new_states.append(new_state)

        return new_states

    def action_cost(self, state, action, state_1):
        station, route = state
        station_1, route_1 = state_1

        if route is None:
            return 0

        if route_1 is None:
            if station == self.goal_station:
                return 0
            return TRANSFER_TIME

        return TRAVEL_TIME + DWELL_TIME

    def heuristic(self, state):
        station = state[0]
        return self.distances_to_goal.get(station, 0)


FAIL = Node(None)
TRAVEL_TIME = 3
DWELL_TIME = 4
TRANSFER_TIME = 5

def expand(problem_inst, node):
    state = node.state
    neighbors = problem_inst.result_actions(state)
    neighbors = list(neighbors)
    expansion = []

    for neighbor in neighbors:
        cost = problem_inst.action_cost(state, neighbor, neighbor) + node.cost
        node_1 = Node(neighbor, node, neighbor, cost)
        expansion.append(node_1)

    return expansion


def best_first_search(problem_inst, evaluation_function):
    root_node = Node(problem_inst.initial_state, None, None, 0)
    frontier = Frontier(evaluation_function)
    frontier.add(root_node)
    reached = {}

    while not frontier.is_empty():
        current_node = frontier.pop()

        if current_node.state in reached and reached[current_node.state] <= current_node.cost:
            continue

        reached[current_node.state] = current_node.cost

        if problem_inst.is_goal(current_node.state):
            return current_node

        for neighbor in expand(problem_inst, current_node):
            frontier.add(neighbor)

    return FAIL


def path_actions(node):
    nodes = []
    while True:
        nodes.append(node)
        if node.parent is None:
            break
        node = node.parent
    nodes.reverse()
    actions = []

    for n in nodes:
        action = n.state
        actions.append(action)
    return actions


def show_result(problem_inst, solution_path):
    for i in range(1, len(solution_path)):
        previous_station, previous_route = solution_path[i - 1]
        station, route = solution_path[i]

        if previous_route is None and route is not None:
            print(f"Tomar: {route}")

        elif previous_route is not None and route is None:
            if station != problem_inst.goal_station:
                print(f"Transbordo en: {station}")


problem = Problem("Portal Tunal", "Portal Norte", routes, graph)

result = best_first_search(
    problem,
    lambda node: node.cost + problem.heuristic(node.state)
)

if result is FAIL:
    print("No se encontro ruta.")
else:
    solution = path_actions(result)
    show_result(problem, solution)
    print()
    print("Duracion:", result.cost, "minutos")
```
